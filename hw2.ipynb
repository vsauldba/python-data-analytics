```python
# 1-Import The Pandas library
import pandas as pd
```

```python
# 2-Load the mtcars dataset from Github
df = pd.read_csv('https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/refs/heads/master/mtcars.csv')
```

```python
# 3-Displays the first 5 rows of the dataset
df.head()
```

```python
# 4-Display the  number of rows and columns
df.shape
```

```python
# 5-Display all Column Names
df.columns
```

```python
# 6-Get Information about the dataset such as: Column Names, Data Types and Missing Values
df.info()
```

```python
# 7-Display Summary Statistics for the numeric Columns
df.describe()
```

```python
# 8-Rename the missing Columns
df.rename(columns={'Unnamed: 0':'car'}, inplace=True)
```

```python
# 9-Displays the 5 rows after renaming the column
df.head()
```

```python
# 10- Create a new dataframe with only the required columns such as: mpg, hp, wt and cyl.
cars = df[['mpg', 'hp', 'wt', 'cyl']]
# Show the first 5 rows of the new dataframe
cars.head()
```

```python
# 10- Rename the "cyl" to "cylinders"
cars = cars.rename(columns={"cyl":"cylinders"})

```

```python
# 11- Display the new column name
cars.head()
```

```python
# 12- Displays Summary statistics for mpg, hp, wt and cylinders
cars.describe()
```

```python
# 13-Calculate the average miles per gallon
cars['mpg'].mean()
```

Comments: The average MPG for the vehicle is approximately **20.09 miles per gallon.**


```python
# 14-Select cars with more than 150 horsepower
high_hp = cars[cars['hp'] > 150]
```

```python
# 15- Displays cars with more than 150 horsepower
high_hp
```

```python
# 16- Display summary statistics for cars with more than 150horsepower
high_hp.describe()
```

```python
# 17- Calculate the average horsepower
cars['hp'].mean()
```

The average horsepower of the vehicle is approximately 146.69 hp.
