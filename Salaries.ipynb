```python
#import pandas
import pandas as pd
```

```python
#Load the dataset from github
df = pd.read_csv("https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/Salaries.csv")
```

```python
# Display The first 5 rows
df.head()
```

```python
#Number of rows and columns
df.shape
```

```python
# Checking the data types
df.dtypes
```

```python
# Count null values in each column
df.isnull().sum()
```

The dataset has no missing values!

```python
df["Unnamed: 0"]
```

```python
#Rename the first column
df.rename(columns={"Unnamed: 0":"id"},inplace=True)
df.head()
```

```python
#The mean of the salaries
df["salary"].mean()
```

The mean salary is: $113,706

```python
#The median of the salalry
df["salary"].median()
```

The median salary is : $107,300

```python
# The median salary by gender
df.groupby("sex")["salary"].median()
```

```python
# Vizualization
# Salary first
df["salary"].plot(
    kind="hist",
    title="Salary frequence among professors",
    xlabel="Salary",
    ylabel="Frequency"

)
```

```python
#
df.groupby("sex")["salary"].median().plot(
    kind="bar",
    title="Median salary by gender",
    xlabel="Gender",
    ylabel="Salary"
)
```

Conclusion: Women have lower median salary than men!

```python
# Check the different ranks
df["rank"].unique()
```

```python
df.groupby("rank")["salary"].median().sort_values(ascending=False).plot(
    kind="bar",
    title="Median salary by rank",
    xlabel="rank",
    ylabel="Salary"
)
```


Conclusion:
In this datasets there three categories of Professors such as : Professors,Associate Professors and Assistant Professors. That Professors have the highest median salary!
